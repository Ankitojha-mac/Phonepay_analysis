```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer('all-MiniLM-L6-v2')

categories = ["Food and Dining",
    "Groceries and Daily Needs",
    "Shopping and E-commerce",
    "Travel and Transport",
    "Bills and Utilities",
    "Health and Medical",
    "Money Transfer and Investment",
    "Entertainment and Leisure",
    "Money Deposited or Received",
    "Education / Bills & Utilities",
    "Housing / Rent"
]

category_embeddings = model.encode(categories, convert_to_tensor=True)


def classify_transaction(text):
    if not isinstance(text, str) or text.strip() == "":
        return "Miscellaneous"
    

    text_embedding = model.encode(text, convert_to_tensor=True)
    
    cosine_scores = util.cos_sim(text_embedding, category_embeddings)

    best_match_idx = cosine_scores.argmax().item()
    return categories[best_match_idx]
```

```python

data['Category'] = data['Transaction Details'].apply(classify_transaction)

data[['Transaction Details', 'Amount', 'Category']].head(10)
```

```python
import re
import spacy
import pandas as pd
from sentence_transformers import SentenceTransformer, util

# 1. Load SpaCy NLP Model and SentenceTransformer
nlp = spacy.load("en_core_web_sm")
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Categories List
categories = ["Food and Dining",
    "Groceries and Daily Needs",
    "Shopping and E-commerce",
    "Travel and Transport",
    "Bills and Utilities",
    "Health and Medical",
    "Money Transfer and Investment",
    "Entertainment and Leisure",
    "Money Deposited or Received",
    "Education / Bills & Utilities",
    "Housing / Rent"
]

category_embeddings = model.encode(categories, convert_to_tensor=True)

# 3. Consolidated & Expanded Keyword Regex Patterns
EDUCATION_PATTERN = r'\b(vishwavidyalaya|college|education|iit|institute|university|school|proudyogiki)\b'
HOUSING_PATTERN = r'\b(hostel|pg|rent|boys hostel|girls hostel)\b'
FOOD_PATTERN = r'\b(swiggy|zomato|cafe|sweets|bakery|restaurant|dhaba|nasta|jalebi|misthan|burger|food|crave|kulfi|icecream|ice cream|canteen|caterar|caterers|restorent)\b'
GROCERY_PATTERN = r'\b(kirana|provision|vegetable|fruit|dairy|store|mart|supermarket)\b'
TRAVEL_PATTERN = r'\b(ixigo|irctc|uts|railway|railways|redbus|bus|metro|uber|ola|rapido)\b'
RECHARGE_PATTERN = r'\b(recharge|bill|mobile|dth|jio|airtel|\bvi\b|bsnl|electricity|gas|broadband|tata play|air fiber|fiber)\b'
SHOPPING_PATTERN = r'\b(ekart|amazon|flipkart|myntra)\b'
HEALTH_PATTERN = r'\b(pharmacy|medical|chemist|hospital|clinic|lab|pathology|diagnostics|apollo|pharmeasy|1mg|netmeds|doctor|dr|lenskart|healthcare|dental|physician|radiology|xray|mri|rehabilitation|therapy|wellness|nutrition|optical|ambulance)\b'

def Smart_Categorize_SpaCy(df):
    results = [None] * len(df)
    texts_to_embed = []
    embed_indices_map = []
    
    for idx, row in df.iterrows():
        text_str = str(row['Transaction Details']).strip()
        txn_type = str(row.get('Type', '')).upper()
        
        # Priority 1: Direct Inflow / Deposit Rules
        if re.search(r'Received from', text_str, re.IGNORECASE) or txn_type == 'CREDIT':
            results[idx] = "Money Deposited or Received"
            continue

        # Priority 2: Specific Merchant & Service Patterns
        if re.search(EDUCATION_PATTERN, text_str, re.IGNORECASE):
            results[idx] = "Education / Bills & Utilities"
            continue
        if re.search(HOUSING_PATTERN, text_str, re.IGNORECASE):
            results[idx] = "Housing / Rent"
            continue
        if re.search(FOOD_PATTERN, text_str, re.IGNORECASE):
            results[idx] = "Food and Dining"
            continue
        if re.search(GROCERY_PATTERN, text_str, re.IGNORECASE):
            results[idx] = "Groceries and Daily Needs"
            continue
        if re.search(TRAVEL_PATTERN, text_str, re.IGNORECASE):
            results[idx] = "Travel and Transport"
            continue
        if re.search(RECHARGE_PATTERN, text_str, re.IGNORECASE):
            results[idx] = "Bills and Utilities"
            continue
        if re.search(SHOPPING_PATTERN, text_str, re.IGNORECASE):
            results[idx] = "Shopping and E-commerce"
            continue
        if re.search(HEALTH_PATTERN, text_str, re.IGNORECASE):
            results[idx] = "Health and Medical"
            continue

        # Priority 3: Fallback Money Transfer / Banking Keyword Patterns
        if re.search(r'Paid to|Transfer|Bank Account|\*+|\b\d{10}\b', text_str, re.IGNORECASE):
            results[idx] = "Money Transfer and Investment"
            continue

        # Priority 4: SpaCy Person Name Detection
        doc = nlp(text_str)
        is_person = any(ent.label_ == "PERSON" for ent in doc.ents)
        if is_person:
            results[idx] = "Money Transfer and Investment"
            continue

        # Priority 5: Queue for Deep Learning Embedding Fallback
        texts_to_embed.append(text_str)
        embed_indices_map.append(idx)
            
    # Deep Learning Cosine Similarity Fallback
    if texts_to_embed:
        embeddings = model.encode(texts_to_embed, convert_to_tensor=True)
        scores = util.cos_sim(embeddings, category_embeddings)
        
        for i, original_idx in enumerate(embed_indices_map):
            max_score, best_idx = scores[i].max(dim=0)
            if max_score.item() < 0.35:
                results[original_idx] = "Money Transfer and Investment"
            else:
                results[original_idx] = categories[best_idx.item()]
                
    return results

# 4. Apply Categorization
data['Category'] = Smart_Categorize_SpaCy(data)

# 5. Check Output
data[['Transaction Details', 'Amount', 'Category']].head(60)
```

```python
x = data["Transaction Details"].fillna("").astype(str).str.lower().str.strip()
y = data["Category"]
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y, test_size = 0.2, random_state = 42 , stratify = y ) 
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(ngram_range =(1,2 ),max_features =5000)
x_train_vec = tfidf.fit_transform(x_train)
x_test_vec = tfidf.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
import joblib
model = LogisticRegression(class_weight='balanced', max_iter=1000)
model.fit(x_train_vec, y_train)
accuracy = model.score(x_test_vec, y_test)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

```python
import joblib
joblib.dump(model, 'transaction_classifier.pkl')
joblib.dump(tfidf, 'tfidf_vectorizer.pkl')
```

```python
import pandas as pd

# Load ground truth or categorized dataframe
data = pd.read_csv('save.csv')

# --- PHASE 3: ANALYTICS ENGINE FUNCTIONS ---

def generate_financial_analytics(df):
    print("="*50)
    print("        AUTOMATED FINANCIAL ANALYTICS REPORT        ")
    print("="*50 + "\n")
    
    # 1. Category-wise Total Expense Breakdown
    print("1. CATEGORY-WISE SPENDING SUMMARY:")
    category_summary = df.groupby('Category')['Amount'].agg(['sum', 'count']).reset_index()
    category_summary.columns = ['Category', 'Total Amount (₹)', 'Transaction Count']
    category_summary['Percentage (%)'] = (category_summary['Total Amount (₹)'] / category_summary['Total Amount (₹)'].sum()) * 100
    category_summary = category_summary.sort_values(by='Total Amount (₹)', ascending=False)
    print(category_summary.to_string(index=False))
    print("\n" + "-"*50 + "\n")
    
    # 2. Monthly Trend Analysis (if 'Month' column exists)
    if 'Month' in df.columns:
        print("2. MONTHLY SPENDING TRENDS:")
        monthly_summary = df.groupby('Month')['Amount'].sum().reset_index()
        monthly_summary.columns = ['Month', 'Total Spent (₹)']
        print(monthly_summary.to_string(index=False))
        print("\n" + "-"*50 + "\n")
        
    # 3. Top 5 Merchants/Receivers
    print("3. TOP 5 MOST FREQUENT MERCHANTS/PAYEES:")
    top_merchants = df['Transaction Details'].value_counts().head(5).reset_index()
    top_merchants.columns = ['Merchant / Payee Details', 'Frequency']
    print(top_merchants.to_string(index=False))
    print("\n" + "="*50)

# Run Analytics Engine
generate_financial_analytics(data)
```
